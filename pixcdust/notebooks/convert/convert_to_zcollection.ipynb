```python
from glob import glob

from pixcdust.converters.zarr import Nc2ZarrConverter
```

# Create a zarr database

```python
pixc = Nc2ZarrConverter(
    path_in=sorted(glob("/tmp/pixc2/**/*.nc", recursive=True)),
    variables=["height", "sig0", "classification"],
)
pixc.database_from_nc(path_out="/tmp/pixc2/pixc_zarr")
```

# Read a zarr database

```python
import datetime

from pixcdust.readers.zarr import ZarrReader
```

```python
pixcr = ZarrReader("/tmp/pixc2/pixc_zarr")
pixcr.read(
    (
        datetime.datetime(
            2023, 4, 6, tzinfo=datetime.datetime.now(datetime.UTC).astimezone().tzinfo
        ),
        datetime.datetime(
            2023, 4, 8, tzinfo=datetime.datetime.now(datetime.UTC).astimezone().tzinfo
        ),
    )
)
pixc
```

```python
gdf_pixc = pixcr.to_geodataframe()
gdf_pixc
```
